# AnimalOrganizer Modes Guide

`AnimalOrganizer` (AO) is responsible for discovering data for a specific animal and grouping it into "days" (one `LongRecordingOrganizer` per day). 

Depending on your data structure, you can configure AO to find data in different ways.

## Discovery Type: Folder vs File

By default, AO looks for **subfolders**. However, you can also tell it to look for **individual files** by providing a `file_pattern`.

### 1. Folder-based Discovery (Default)
AO finds subfolders and treats each one as a separate session or part of a session.
- **`nest`**: Expects `base / animal_id / day_folder / content`.
- **`concat`**: Expects `base / session_folder / content`, where `session_folder` contains `animal_id`.

### 2. File-based Discovery (New)
If you provide a `file_pattern` (e.g., `*.nwb`), AO will glob for files and then use the **dates in the filenames** to group them into separate days.
- **Useful for**: A flat folder containing many NWB files for the same animal across multiple days.

## Mode Reference

| Mode | Search Logic | Best For... |
|------|--------------|-------------|
| **`nest`** | Subfolders of an animal folder | Standard daily folder structure |
| **`concat`** | Folders/Files with animal ID in name | Continuous recordings split across files/folders |
| **`noday`** | Same as `concat` | When you want exactly **one** recording for the whole time range |
| **`base`** | Use the base folder directly | Single session, no sub-discovery needed |

## Configuration Examples

### Typical Nested Folders
```yaml
war_generation:
  mode: "nest"
  # file_pattern is None (default)
```

### Flat NWB Files spanning multiple days
```yaml
war_generation:
  mode: "concat"
  file_pattern: "*.nwb"
  lro_kwargs:
    mode: "si"
```

## Programmatic Usage Example

````python
from neurodent import AnimalOrganizer

# File-based discovery (e.g. for NWB files)
ao = AnimalOrganizer(
    base_folder_path="/path/to/data",
    anim_id="Animal1",
    mode="concat",
    file_pattern="*.nwb",
    lro_kwargs={"mode": "si"}
)

print(f"Found {len(ao.long_recordings)} days for animal: {ao.anim_id}")
````
